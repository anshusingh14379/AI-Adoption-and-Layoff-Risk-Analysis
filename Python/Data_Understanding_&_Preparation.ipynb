### Range Validation
- It is required to check the validation rrule of percentage--->percentage should not > 100 or < 0

```python
percent_columns = ['routine_task_percentage',
                  'creativity_requirement', 
                  'human_interaction_level',
                   'tasks_automated_percentage']
for col in percent_columns:
    invalid = df[(df[col]<0)|(df[col]>100)]   # ---> check is there any value not present in the dataset above 100 or below 0, because if present it violet the percentage validation
    print(f'{col}: {len(invalid)} invalid records')
```

### Logical Consistency
- It is important to check logically consistency which is not possible in actually.

```python
# Lets take a example suppose in my dataset any of employee age is given like 25 and their years of experience should be given like 20---> it is logically incorrect data

invalid_experience = df[df['years_of_experience'] > (df['age']-18)]

df.groupby("job_level")["years_of_experience"].describe()
df.groupby('ai_adoption_level')['ai_usage_hours_per_week'].describe()
```
